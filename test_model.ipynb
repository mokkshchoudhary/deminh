```python
import ollama
```

```python
question = """A company reports the following:
Revenue in 2019: 4,210 million
Revenue in 2018: 3,890 million

What was the change in revenue from 2018 to 2019?"""

response = ollama.chat(
    model="qwen3:8b",
    messages=[{"role": "user", "content": question}]
)

print(response["message"]["content"])
```

```python
import time
import ollama

question = "The revenue in 2018 was 3,890 million and in 2019 was 4,210 million. What is the change in revenue?"

start = time.time()
response = ollama.chat(
    model="qwen3:4b",
    messages=[{"role": "user", "content": question + " /no_think"}],
    options={"num_ctx": 4096}
)
elapsed = time.time() - start

print(f"Took {elapsed:.1f} seconds")
print(response["message"]["content"])
```
